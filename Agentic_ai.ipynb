```python
import os
from dotenv import load_dotenv

load_dotenv()
print(os.getenv("AZURE_OPENAI_GPT_4O_DEPLOYMENT_NAME"))
```

```python
from agents import (
    Agent,
    OpenAIChatCompletionsModel,
    Runner
)
from openai import AsyncAzureOpenAI
```

```python
agent_client = AsyncAzureOpenAI(
    api_key = os.getenv("AZURE_OPENAI_GPT_4O_API_KEY"),
    api_version = os.getenv("AZURE_OPENAI_GPT_4O_API_VERSION"),
    azure_endpoint = os.getenv("AZURE_OPENAI_GPT_4O_ENDPOINT"),
    azure_deployment = os.getenv("AZURE_OPENAI_GPT_4O_DEPLOYMENT_NAME")
)
```

```python
model_name = os.getenv("AZURE_OPENAI_GPT_4O_DEPLOYMENT_NAME")
model_name
```

```python
custom_agent = Agent(
    name = "My custom Agent",
    instructions = "You are a conversational AI assistant, Answer the user's questions based on your knowledge",
    model = OpenAIChatCompletionsModel(model = model_name, openai_client = agent_client),
)
```

```python
result = await Runner.run(
    custom_agent,
    input = "hey hi",
)
```

```python
result.final_output
```

```python
result = await Runner.run(
    custom_agent,
    input = "Write a small blog with headings and bulletpoints on the topic 'AI in Healthcare'",
)
```
