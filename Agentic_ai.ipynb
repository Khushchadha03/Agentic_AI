```python
from dataclasses import dataclass

@dataclass
class UserInfo:
    user_name: str
    user_api_key: str
```

```python
from agents import function_tool, RunContextWrapper
import datetime


@function_tool
def get_weather(wrapper: RunContextWrapper[UserInfo], city_name: str)-> dict:
    """
    Fetch current weather information for a given city using OpenWeatherMap API.

    Args:
        city_name(str): Name of the city(e.g., "London").
    
    Returns:
        dict: Weather information plus metadata{called_at, agent, input_params, result}.
    """

    called_at = datetime.datetime.utcnow().isoformat() + 'Z'
    input_params = {"city_name": city_name}
    agent = 'Travel_agent'
    results = {
        "city" : city_name,
        "temperature" : 18,
        "description" : "Party Cloudy",
        "humidity" : 65,
        "wind_speed" : 12
    }

    api_key = wrapper.context['user_api_key']

    if api_key == "abcdefgh":
        return{
            "called_at": called_at,
            "agent": agent,
            "input_params": input_params,
            "result": results
        }
    else:
        return{"error": "Invalid API key"}
```

```python
custom_agent = Agent[UserInfo](
    name = "Travel Agent",
    instructions = "You are a travelling AI assistant, Answer the user's questions based on your knowledge and use the tools provided",
    model = OpenAIChatCompletionsModel(model = model_name, openai_client = agent_client),
    tools = [get_weather]
)
```

```python
messages = []

my_user_info ={
    "user_name": "Khush Chadha",
    "user_api_key": "abcdefgh"
}

async def ask_agent(query: str):
    global messages
    input_msg = [{"role": "user", "content": query}]
    result = await Runner.run(custom_agent, input = messages + input_msg, context = my_user_info)
    messages = result.to_input_list()
    display(Markdown(result.final_output))
```
